```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import os
df = pd.read_csv("data/clean_data.csv")

os.makedirs("reports/figures", exist_ok=True)

colors = ["#4472C4", "#ED7D31", "#70AD47", "#A5A5A5"]

print(df.head())
print(df.shape)
```

```python
department_count = df["Department"].value_counts()

plt.figure(figsize=(8, 5))

plt.bar(
    department_count.index,
    department_count.values,
    color=colors
)

plt.title("Employee Count by Department")
plt.xlabel("Department")
plt.ylabel("Number of Employees (Count)")
plt.legend(["Employees"])

plt.tight_layout()
plt.savefig("reports/figures/bar_chart.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.hist(
    df["Salary"].dropna(),
    bins=10,
    color="#4472C4",
    edgecolor="black"
)

plt.title("Salary Distribution")
plt.xlabel("Salary (INR)")
plt.ylabel("Number of Employees (Count)")
plt.legend(["Salary"])

plt.tight_layout()
plt.savefig("reports/figures/histogram.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.scatter(
    df["Age"],
    df["Salary"],
    color="#ED7D31",
    label="Employees"
)

plt.title("Age vs Salary")
plt.xlabel("Age (Years)")
plt.ylabel("Salary (INR)")
plt.legend()

plt.tight_layout()
plt.savefig("reports/figures/scatter_plot.png")
plt.show()
```

```python
salary_sorted = df["Salary"].dropna().sort_values().reset_index(drop=True)

plt.figure(figsize=(8, 5))

plt.plot(
    salary_sorted.index + 1,
    salary_sorted.values,
    color="#70AD47",
    marker="o",
    label="Salary"
)

plt.title("Sorted Employee Salary Trend")
plt.xlabel("Employee Rank (Sorted)")
plt.ylabel("Salary (INR)")
plt.legend()

plt.tight_layout()
plt.savefig("reports/figures/line_chart.png")
plt.show()
```

```python
plt.figure(figsize=(8, 5))

plt.boxplot(
    df["Salary"].dropna(),
    patch_artist=True
)

plt.title("Employee Salary Distribution - Box Plot")
plt.xlabel("Employee Salary")
plt.ylabel("Salary (INR)")

plt.tight_layout()
plt.savefig("reports/figures/box_plot.png")
plt.show()
```
